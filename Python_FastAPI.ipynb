Task 2: Build user input workflow for intern details capture

Clean text inputs by removing extra spaces and applying proper casing

Create Intern object from user input


```python
class Intern:
    def __init__(self, name, role):
        self.name = name
        self.role = role
        self.daily_hours = []
        self.daily_tasks = []

    def display(self):
        print("Intern Details:")
        print("Name :", self.name)
        print("Role :", self.role)


def clean_text(text):
    text = text.strip()
    text = " ".join(text.split())
    return text.title()


def capture_intern_details():
    name_input = input("Enter intern name: ")
    role_input = input("Enter intern role: ")

    name = clean_text(name_input)
    role = clean_text(role_input)

    return Intern(name, role)


intern = capture_intern_details()
intern.display()
```

Task 3: Implement daily activity logging loop

Prompt user for number of days to record

For each day: capture hours worked and task description

Store hours and tasks in separate lists


```python
num_days = int(input("Enter number of days: "))

for day in range(1, num_days + 1):
    print(f"Day {day}")

    hours = float(input("Enter hours worked: "))

    tasks_input = input("Enter tasks done with commas): ")
    tasks = tasks_input.split(",")

    cleaned_tasks = []
    for task in tasks:
        cleaned_tasks.append(task.strip())

    intern.daily_hours.append(hours)
    intern.daily_tasks.append(cleaned_tasks)


print("Logged Daily Information:")

for i in range(num_days):
    print(f"Day {i+1}")
    print(" Hours:", intern.daily_hours[i])
    print(" Tasks:")
    for task in intern.daily_tasks[i]:
        print("  -", task)
```

Task 4: Add input validation for work hours

```python
def valid_hours():
    while True:
        try:
            hours = float(input("Enter hours worked: "))

            if hours < 0:
                raise ValueError

            return hours

        except ValueError:
            print("Invalid input. Please enter a non-negative number.")
```

Task 5: Create business logic functions

Function to calculate total working hours

Function to calculate average working hours

Function to generate formatted intern summary

```python
def calculate_total_hours(daily_hours):
    return sum(daily_hours)


def calculate_average_hours(daily_hours):
    if len(daily_hours) == 0:
        return 0
    return sum(daily_hours) / len(daily_hours)

```

```python
# function to generate intern summary 
def generate_intern_summary(intern):
    total_hours = calculate_total_hours(intern.daily_hours)
    average_hours = calculate_average_hours(intern.daily_hours)

    summary = f"""
Intern Summary
--------------
Name          : {intern.name}
Role          : {intern.role}
Total Hours   : {total_hours}
Average Hours : {average_hours:.2f}

Daily Tasks:
"""

    for day, tasks in enumerate(intern.daily_tasks, start=1):
        summary += f"\nDay {day}:\n"
        for task in tasks:
            summary += f"  - {task}\n"

    return summary
```

```python
print(generate_intern_summary(intern))
```

Task 6: Implement JSON data persistence

Convert intern object to dictionary structure

Write intern data to JSON file

Read stored data back from JSON for verification

```python
import json

# Intern object to dictionary
def intern_to_dict(intern):
    return {
        "name": intern.name,
        "role": intern.role,
        "daily_hours": intern.daily_hours,
        "daily_tasks": intern.daily_tasks
    }

# Save intern data to JSON file
def save_intern_to_json(intern, filename="intern_data.json"):
    with open(filename, "w") as file:
        json.dump(intern_to_dict(intern), file, indent=4)
    print(f"Intern data saved to {filename} successfully!")

```

```python

# Load intern data from JSON file
def load_intern_from_json(filename="intern_data.json"):
    with open(filename, "r") as file:
        data = json.load(file)
    
    # Reconstruct Intern object
    loaded_intern = Intern(data["name"], data["role"])
    loaded_intern.daily_hours = data["daily_hours"]
    loaded_intern.daily_tasks = data["daily_tasks"]
    return loaded_intern


# Save your intern object
save_intern_to_json(intern)

# Load it back for verification
loaded_intern = load_intern_from_json()
print("Loaded Intern from JSON:")
print(generate_intern_summary(loaded_intern))
```

#### Tasks Day-2 (Multi-Intern Console System with Status Tracking)

Task 1: Enhance Intern class for multi-intern support

Add attributes: intern_id, status (Active/Inactive/Completed)

Set default status to Active for new interns

```python
class Intern:
    id_counter = 1

    def __init__(self, name, role, status="Active"):
        self.intern_id = "INT" + str(Intern.id_counter)
        Intern.id_counter += 1

        self.name = name
        self.role = role
        self.status = status
        self.activities = []

    def display_info(self):
        print(f"Intern ID  : {self.intern_id}")
        print(f"Name       : {self.name}")
        print(f"Role       : {self.role}")
        print(f"Status     : {self.status}")
```

Task 2: Build Add Intern workflow

Prompt for name and role with input cleaning

Generate unique intern ID (INT001, INT002, etc.)

Store intern in collection (list of dictionaries)

Persist to JSON file after each addition

```python
# Helper Function

import json

filename = "intern_data.json"

# Load interns from JSON
def load_interns():
    try:
        with open(filename, "r") as file:
            return json.load(file)
    except FileNotFoundError:
        return []

# Save interns to JSON
def save_interns(interns):
    with open(filename, "w") as file:
        json.dump(interns, file, indent=4)
```

```python
def clean_text(text):
        text = text.strip()
        text = " ".join(text.split())
        return text.title()

def add_intern():
    interns = load_interns()  # load existing interns from JSON

    name = clean_text(input("Enter intern name: "))
    role = clean_text(input("Enter intern role: "))

    # Generate unique ID
    counter = len(interns) + 1
    intern_id = f"INT{counter:03d}"

    intern_dict = {
        "id": intern_id,
        "name": name,
        "role": role,
        "status": "Active",
        "activities": []
    }

    interns.append(intern_dict)
    save_interns(interns)  # persist to JSON
    print(f"Intern {name} added successfully with ID {intern_id}!")
```

Task 3: Implement Add Daily Activity workflow

Prompt for intern ID and validate existence

Check intern status before allowing activity entry

Allow activity only if status is Active

Prevent activity if status is Inactive or Completed

Append hours and tasks to intern record

Update JSON storage

```python
def add_daily_activity():
    interns = load_interns()

    intern_id = input("Enter Intern ID: ").strip().upper()

    intern = None
    for i in interns:
        if i["id"] == intern_id:
            intern = i
            break

    if intern is None:
        print("Invalid Intern ID.")
        return

    # Check intern status
    if intern["status"] != "Active":
        print("Activity not allowed. Intern status is:", intern["status"])
        return

    # Add daily activity
    try:
        hours = float(input("Enter hours worked today: "))
    except ValueError:
        print("Invalid hours input.")
        return

    task = input("Enter task description: ").strip()

    intern.setdefault("activities", []).append({
        "hours": hours,
        "task": task
    })

    save_interns(interns)
    print("Daily activity added successfully!")
```

Task 4: Create Update Intern Status feature

Allow status changes: Active → Inactive → Completed

Reflect status changes in JSON immediately

Enforce status-based access control


```python
def update_intern_status():
    interns = load_interns()

    intern_id = input("Enter Intern ID to update status: ").strip().upper()

    intern = None
    for i in interns:
        if i["id"] == intern_id:
            intern = i
            break

    if intern is None:
        print("Invalid Intern ID.")
        return

    current_status = intern["status"]
    print("Current Status:", current_status)

    # Decide next status
    if current_status == "Active":
        new_status = "Inactive"

    elif current_status == "Inactive":
        new_status = "Completed"

    else:
        print("Status change not allowed. Intern already Completed.")
        return

    # Update status
    intern["status"] = new_status
    save_interns(interns)

    print(f"Status updated to: {new_status}")
    print("Status change saved to JSON.")

    # Access control message
    if new_status != "Active":
        print("Access Update: Daily activity entry is now blocked.")
```

Task 5: Build View Intern Summary feature

Retrieve intern by ID

Display intern details including current status

Calculate and show total/average hours

Display task history and most recent activity

```python
def view_intern_summary():
    interns = load_interns()

    intern_id = input("Enter Intern ID to view summary: ").strip().upper()

    intern = None
    for i in interns:
        if i["id"] == intern_id:
            intern = i
            break

    if intern is None:
        print("Invalid Intern ID.")
        return

    print("\n======= INTERN SUMMARY =======")
    print("ID     :", intern["id"])
    print("Name   :", intern["name"])
    print("Role   :", intern["role"])
    print("Status :", intern["status"])

    activities = intern.get("activities", [])

    total_hours = sum(a["hours"] for a in activities)
    avg_hours = total_hours / len(activities) if activities else 0

    print("Total Hours Worked   :", total_hours)
    print("Average Hours / Day  :", round(avg_hours, 2))
    print("------------------------------")

    if activities:
        print("Task History:")
        for idx, act in enumerate(activities, start=1):
            print(f"Day {idx}: {act['hours']} hrs - {act['task']}")

        last_activity = activities[-1]
        print("\nMost Recent Activity:")
        print(f"{last_activity['hours']} hrs - {last_activity['task']}")
    else:
        print("No activity recorded yet.")
```

Task 6: Implement Overall Statistics feature

Count total number of interns

Count active interns separately

Calculate average hours across all interns

Identify top performer by total hours worked


```python
def overall_statistics():
    interns = load_interns()

    if not interns:
        print("No interns found.")
        return

    # Count interns
    total_interns = len(interns)
    active_interns = 0

    total_hours_all = 0
    top_performer = None
    max_hours = 0

    for intern in interns:
        if intern["status"] == "Active":
            active_interns += 1

        activities = intern.get("activities", [])
        intern_hours = sum(a["hours"] for a in activities)

        total_hours_all += intern_hours

        if intern_hours > max_hours:
            max_hours = intern_hours
            top_performer = intern

    average_hours = total_hours_all / total_interns if total_interns > 0 else 0

    print("\n======= OVERALL INTERN STATISTICS =======")
    print(f"Total Interns        : {total_interns}")
    print(f"Active Interns       : {active_interns}")
    print(f"Average Hours/Intern : {average_hours:.2f}")

    if top_performer:
        print(f"Top Performer        : {top_performer['name']} ({max_hours} hrs)")
    else:
        print("Top Performer        : No activities recorded.")
```

Task 7: Design menu-driven interface

Create main menu with all operations

Implement input-driven navigation

Add exit option to terminate program


```python
def main_menu():
    while True:
        print("\n========== INTERN MANAGEMENT SYSTEM ==========")
        print("1. Add Intern")
        print("2. Add Daily Activity")
        print("3. Update Intern Status")
        print("4. View Intern Summary")
        print("5. Overall Statistics")
        print("6. Exit")

        choice = input("Enter your choice (1-6): ").strip()

        if choice == "1":
            add_intern()

        elif choice == "2":
            add_daily_activity()

        elif choice == "3":
            update_intern_status()

        elif choice == "4":
            view_intern_summary()

        elif choice == "5":
            overall_statistics()

        elif choice == "6":
            print("Exiting Program. Goodbye!")
            break

        else:
            print("Invalid choice! Please select between 1 and 6.")


if __name__ == "__main__":
    main_menu()

```
